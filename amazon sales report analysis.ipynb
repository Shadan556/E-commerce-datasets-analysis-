```python
# checking zero order  status
amazon[amazon["Amount"] == 0]["Status"].value_counts()
```

```python
#creating pivot tables for total sales by state
state_sales_pivot = pd.pivot_table(
    data = amazon, # data frame
    values="Amount", #numeric column to calculate
    index = "ship-state", # all rows grouby state
    aggfunc = "sum" # add all sales amounts
)
#sort high sales to low sales
state_sales_pivot = state_sales_pivot.sort_values(
    by ="Amount", ascending = False
)
state_sales_pivot.head(20) #will show top 20 satates
```

```python
# charts for the above result
state_sales_pivot.head(20).plot(
    kind = "bar", figsize = (12,6)
)
plt.title("Top 20 States By Revenue")
plt.xlabel("State")
plt.ylabel("total Sales Amount")

plt.xticks(rotation = 45 , ha = "right")
plt.tight_layout()
plt.show()
```

```python
# creating cross tab of state vs order status
state_status_table = pd.crosstab(
    amazon["ship-state"],
    amazon["Status"]
    
)
state_status_table["total orders"] = state_status_table.sum(axis=1)
state_status_table.head(20)
```

```python
# chart for above explanation for better explanation
state_status_table[["Shipped","Cancelled"]].head(20).plot(
    kind = "bar", figsize = (12,6)
)
plt.title("Shipped vs Cancelled Orders by State")
plt.xlabel("State")
plt.ylabel("Number of Orders ")
plt.xticks(rotation = 45 , ha = "right")
plt.tight_layout # matplotlib layout formating command
plt.show()
```

```python
# now we group data by product category and calculate total revenue
category_sales = amazon.groupby("Category")["Amount"].sum()
#sort
category_sales = category_sales.sort_values(ascending = False)
category_sales.head(9)
```

```python
category_sales.head(9).plot(
    kind="bar",figsize = (12,6)
    )
plt.title("Top 9 Category by Revenue")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation = 45 , ha = "right")
plt.tight_layout()
plt.show()
```
